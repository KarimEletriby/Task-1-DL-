```python
import numpy as np

# -----------------------------
# 1. Input sentence (tokens)
# -----------------------------
tokens = ["What", "are", "the", "symptoms", "of", "diabetes", "?"]
seq_len = len(tokens)

# -----------------------------
# 2. Create random embeddings
# (In practice: use pretrained embeddings)
# -----------------------------
d_model = 8  # embedding size (small for demo)

np.random.seed(42)
X = np.random.rand(seq_len, d_model)  # shape: (7, 8)

# -----------------------------
# 3. Initialize weight matrices
# -----------------------------
W_Q = np.random.rand(d_model, d_model)
W_K = np.random.rand(d_model, d_model)
W_V = np.random.rand(d_model, d_model)

# -----------------------------
# 4. Compute Q, K, V
# -----------------------------
Q = X @ W_Q   # (7, 8)
K = X @ W_K   # (7, 8)
V = X @ W_V   # (7, 8)

# -----------------------------
# 5. Compute Attention Scores
# -----------------------------
dk = d_model
scores = (Q @ K.T) / np.sqrt(dk)   # (7, 7)

# -----------------------------
# 6. Apply Softmax
# -----------------------------
def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)

attention_weights = softmax(scores)  # (7, 7)

# -----------------------------
# 7. Compute Final Output
# -----------------------------
output = attention_weights @ V  # (7, 8)

# -----------------------------
# 8. Print Results
# -----------------------------
print("Tokens:", tokens)
print("\nAttention Scores:\n", scores)
print("\nAttention Weights (after softmax):\n", attention_weights)
print("\nSelf-Attention Output:\n", output)
```
